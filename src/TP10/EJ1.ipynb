```python

import numpy_financial as npf

# Parámetros globales
r = 0.01  # Tasa efectiva mensual (1%)

# a. Plan original (12 meses, $1000/mes)
flujos_a = [-1000] * 12
flujos_a[-1] += 18000  # Ingreso neto en mes 12
vp_inv_a = sum(1000 / (1 + r)**t for t in range(1, 13))
npv_a = npf.npv(r, [0] + flujos_a)
rent_a = (npv_a / vp_inv_a) * 100

# b. Extensión 3 meses (15 meses, $1000/mes)
flujos_b = [-1000] * 15
flujos_b[-1] += 18000
vp_inv_b = sum(1000 / (1 + r)**t for t in range(1, 16))
npv_b = npf.npv(r, [0] + flujos_b)
rent_b = (npv_b / vp_inv_b) * 100

# d. Retraso de 6 meses (18 meses, $1000/mes)
flujos_d = [-1000] * 18
flujos_d[-1] += 18000
vp_inv_d = sum(1000 / (1 + r)**t for t in range(1, 19))
npv_d = npf.npv(r, [0] + flujos_d)
rent_d = (npv_d / vp_inv_d) * 100

# e. Gestión profesional (+5% costo = $1050/mes, 12 meses)
flujos_e = [-1050] * 12
flujos_e[-1] += 18000
vp_inv_e = sum(1050 / (1 + r)**t for t in range(1, 13))
npv_e = npf.npv(r, [0] + flujos_e)
rent_e = (npv_e / vp_inv_e) * 100

# f. Desafío (Costo máximo para 15m manteniendo NPV del plan original)
factor_15 = sum(1 / (1 + r)**t for t in range(1, 16))
vp_ingreso_15 = 18000 / (1 + r)**15
c_max = (vp_ingreso_15 - npv_a) / factor_15

print(f'a) VPN: ${npv_a:.2f} | Rentabilidad: {rent_a:.2f}%')
print(f'b) VPN: ${npv_b:.2f} | Rentabilidad: {rent_b:.2f}%')
print(f'd) VPN: ${npv_d:.2f} | Rentabilidad: {rent_d:.2f}%')
print(f'e) VPN: ${npv_e:.2f} | Rentabilidad: {rent_e:.2f}%')
print(f'f) Costo mensual máximo aceptable: ${c_max:.2f}')
```
