```python
# Importar librerías
import numpy as np
import matplotlib.pyplot as plt

# Fórmulas
def esfuerzo(S):
    return 8 * (S ** 0.95)

def tiempo(E):
    return 2.4 * (E ** 0.33)

# Intervalos
S = np.linspace(0, 10000, 1000)   # Tamaños del proyecto
E = esfuerzo(S)

E_intervalo = np.linspace(1, 500, 1000)  # Esfuerzos
td = tiempo(E_intervalo)

# Gráficos
plt.figure(figsize=(12,5))

# Gráfico 1: Esfuerzo vs Tamaño
plt.subplot(1,2,1)
plt.plot(S, E, color='blue')
plt.title('Esfuerzo (E) vs Tamaño (S)')
plt.xlabel('Tamaño del proyecto (S)')
plt.ylabel('Esfuerzo (E)')
plt.grid(True)

# Gráfico 2: Tiempo vs Esfuerzo
plt.subplot(1,2,2)
plt.plot(E_intervalo, td, color='green')
plt.title('Tiempo (td) vs Esfuerzo (E)')
plt.xlabel('Esfuerzo (E)')
plt.ylabel('Tiempo calendario (td)')
plt.grid(True)

plt.tight_layout()
plt.show()
```
